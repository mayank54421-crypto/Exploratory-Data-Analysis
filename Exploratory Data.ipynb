```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside Jupyter Notebook
%matplotlib inline

# Set visualization style
sns.set_style("whitegrid")
```

```python
df = sns.load_dataset("titanic")

print("Dataset loaded successfully!")

df.head()

df.tail()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print(df.columns)

df.info()
```

```python
df.describe()

df.describe(include="object")

print("Average age:", df["age"].mean())
print("Median age:", df["age"].median())
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())

print("Average fare:", df["fare"].mean())
```

```python
df.isnull().sum()

plt.figure(figsize=(10, 5))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values in Titanic Dataset")
plt.show()

missing_percentage = df.isnull().mean() * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
df["age"] = df["age"].fillna(df["age"].median())

df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print(df.shape)
```

```python
print(df["survived"].value_counts())

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="survived")

plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
survival_rate = df["survived"].mean() * 100

print(f"Overall Survival Rate: {survival_rate:.2f}%")
```

```python
print(df["sex"].value_counts())

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="sex", hue="survived")

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()

gender_survival = df.groupby("sex")["survived"].mean() * 100

print(gender_survival)
```

```python
print(df["class"].value_counts())

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="class", hue="survived")

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()

class_survival = df.groupby("class", observed=True)["survived"].mean() * 100

print(class_survival)
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(df["age"], bins=30, kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()

plt.figure(figsize=(8, 5))

sns.boxplot(x=df["age"])

plt.title("Age Distribution Using Box Plot")
plt.xlabel("Age")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(df["fare"], bins=30, kde=True)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()

plt.figure(figsize=(8, 5))

sns.boxplot(x=df["fare"])

plt.title("Fare Box Plot")
plt.xlabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(data=df, x="survived", y="age")

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df, x="class", hue="sex")

plt.title("Passenger Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="class",
    y="survived",
    hue="sex",
    estimator="mean"
)

plt.title("Survival Rate by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
df["family_size"] = df["sibsp"] + df["parch"] + 1

df[["sibsp", "parch", "family_size"]].head()

plt.figure(figsize=(10, 5))

sns.countplot(data=df, x="family_size")

plt.title("Passenger Distribution by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="family_size",
    y="survived",
    estimator="mean"
)

plt.title("Survival Rate by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Survival Rate")

plt.show()
```

```python
numeric_columns = df.select_dtypes(include=np.number)

correlation_matrix = numeric_columns.corr()

print(correlation_matrix)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="embarked", hue="survived")

plt.title("Survival by Embarkation Port")
plt.xlabel("Embarkation Port")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="alone", hue="survived")

plt.title("Survival Based on Travelling Alone")
plt.xlabel("Travelled Alone")
plt.ylabel("Number of Passengers")

plt.show()

alone_survival = df.groupby("alone", observed=True)["survived"].mean() * 100

print(alone_survival)
```

```python
factors = ["sex", "class", "age", "fare", "family_size", "alone"]

for factor in factors:
    print("\n---", factor, "---")
    
    if df[factor].dtype == "object" or str(df[factor].dtype) == "category" or df[factor].dtype == "bool":
        print(df.groupby(factor, observed=True)["survived"].mean())
    else:
        print(df[[factor, "survived"]].corr())
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Passengers",
        "Average Age",
        "Median Age",
        "Average Fare",
        "Survival Rate"
    ],
    "Value": [
        len(df),
        df["age"].mean(),
        df["age"].median(),
        df["fare"].mean(),
        df["survived"].mean() * 100
    ]
})

summary
```
